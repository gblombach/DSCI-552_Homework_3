### (c) Feature Extraction

Classification of time series usually needs extracting features from them. In this problem, we focus on time-domain features.

#### i. Research

Research what types of time-domain features are usually used in time series classification and list them (examples are minimum, maximum, mean, etc).

My research included:
- Fulcher, B. D., & Jones, N. S. (2014). *Highly comparative feature-based time-series classification. IEEE Transactions on Knowledge and Data Engineering*, 26(12), 3026-3037. https://doi.org/10.1109/TKDE.2014.2316504
- Lei, Y., & Wu, Z. (2020). Time series classification based on statistical features. *EURASIP Journal on Wireless Communications and Networking, 2020*(46), 1-13, https://link.springer.com/content/pdf/10.1186/s13638-020-1661-4.pdf
- Bagnall, A., Lines, J., Bostrom, A., Large, J., & Keogh, E. (2017). *The great time series classification bake off: A review and experimental evaluation of recent algorithmic advances. Data Mining and Knowledge Discovery, 31*(3), 606-660.

My research revealed these common time-domain features:

1. mean: the average value of the series
2. spread: the degree of variability
   - range
   - variance
   - interquartile
   - standard deviation
3. trend: long term direction of series
4. skewness: measure of the asymetry of the distribution
5. kurtosis:"tailedness" or peakedness of a distribution 
6. seasonality: regular patterns that repeat at fixed intervals, such as daily, weekly, monthly, or yearly cycle
7. periodicity: repeating cycles with a characteristic period, not correspond to calendar intervals
8. serial correlation: correlation between observations and their past values
9. chaos: complex deterministic behavior that appears random and exhibits extreme sensitivity to initial conditions
10. nonlinearity: when relationships within the time series cannot be adequately described by linear models
11. self-similarity: patterns that appear similar across different time scales, like fractal and long-memory processes
